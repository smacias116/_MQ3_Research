```python
# --- Method 1: Create 4 individual DataFrames ---
df_v1 = df[df["Video Session"] == 1]
df_v2 = df[df["Video Session"] == 2]
df_v3 = df[df["Video Session"] == 3]
df_v4 = df[df["Video Session"] == 4]
```

### TABLES IN __DFS{}__ DICTIONARY:

```python
# CHECKING MANUAL CONTROL RESULTS:
total_seconds_v1 = (df_v1["Video Session"] == 1).sum()
total_seconds_v2 = (df_v2["Video Session"] == 2).sum()
total_seconds_v3 = (df_v3["Video Session"] == 3).sum()
total_seconds_v4 = (df_v4["Video Session"] == 4).sum()

minutes_v1 = total_seconds_v1  // 60
seconds_v1 = total_seconds_v1  % 60
minutes_v2 = total_seconds_v2  // 60
seconds_v2 = total_seconds_v2  % 60
minutes_v3 = total_seconds_v3  // 60
seconds_v3 = total_seconds_v3  % 60
minutes_v4 = total_seconds_v4  // 60
seconds_v4 = total_seconds_v4  % 60

print(f"v1 Total time: {minutes_v1}:{seconds_v1:02d}")
print(f"v2 Total time: {minutes_v2}:{seconds_v2:02d}")
print(f"v3 Total time: {minutes_v3}:{seconds_v3:02d}")
print(f"v4 Total time: {minutes_v4}:{seconds_v4:02d}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

# Pack your 4 individual DataFrames into an iterable collection with their labels
video_list = [
    (1, df_v1),
    (2, df_v2),
    (3, df_v3),
    (4, df_v4)
]

for key, df in video_list:
    df.columns = df.columns.str.strip()
    x = df["Elapsed (s)"]
    y = df["Accumulated Power (mWh)"]
    
    # Plot the line and capture the line object to get its color
    line, = plt.plot(x, y, label=f"V{key}") # Changed to V1, V2, etc. for a cleaner legend
    
    # Find the index and coordinates of the highest y-value
    max_idx = y.idxmax()
    max_x = x.loc[max_idx]
    max_y = y.loc[max_idx]
    
    # Calculate max value / max elapsed time
    ratio = max_y / max_x if max_x != 0 else 0
    
    # Add a matching dot at the highest point
    plt.scatter(max_x, max_y, color=line.get_color(), zorder=5)
    
    # Annotate with the value and ratio on the same line
    plt.annotate(
        f"{max_y:.1f} ({ratio:.2f})", 
        (max_x, max_y),
        textcoords="offset points",
        xytext=(5, -10),      # 5 points to the right, 10 points down
        ha='left',            # Aligned to the left so text extends to the right
        va='center',          # Vertically centered relative to the dot
        color=line.get_color(), 
        fontsize=9, 
        weight='bold'
    )

# Add titles, labels, and legend
plt.title("Consumed Power over Time")
plt.xlabel("Elapsed Time (s)")
plt.ylabel("Accumulated Power (mWh)")
plt.legend()
plt.grid(True)

plt.show()
```
